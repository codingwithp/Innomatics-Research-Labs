```python
from langchain_community.llms import HuggingFacePipeline
from transformers import pipeline

# correct pipeline task
pipe = pipeline(
    "text-generation",   # ✅ FIXED
    model="google/flan-t5-base"
)

llm = HuggingFacePipeline(pipeline=pipe)

print(llm.invoke("Explain AI in one sentence"))
```

```python
from langchain_core.prompts import PromptTemplate

template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms"
)

prompt = template.format(topic="LangChain")
print(llm.invoke(prompt))
```

```python
from langchain_classic.chains import LLMChain

chain = LLMChain(llm=llm, prompt=template)

print(chain.run("Machine Learning"))
```

```python
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import ConversationChain

memory = ConversationBufferMemory()

conversation = ConversationChain(
    llm=llm,
    memory=memory
)

conversation.run("Hi, my name is Alex")
print(conversation.run("What is my name?"))
```

```python
def calculator_tool(expression):
    return eval(expression)

query = "45 * 3"
result = calculator_tool(query)

print(f"Result: {result}")
```

```python
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings()

texts = [
    "LangChain is a framework for LLM applications",
    "AI is transforming industries"
]

vectorstore = FAISS.from_texts(texts, embeddings)

query = "What is LangChain?"
docs = vectorstore.similarity_search(query)

print(docs)
```
